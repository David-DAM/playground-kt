```kotlin
%use dataframe, kandy
```

```kotlin
// Generate list of random values
val values = List(50) { kotlin.random.Random.nextDouble(0.0, 100.0) }
val indices = values.indices.toList()

// Create DataFrame
val df = dataFrameOf(
    "index" to indices,
    "value" to values
)

df
```

```kotlin
// Visualize with line chart
df.plot {
    line {
        x(index)
        y(value)
    }
    layout {
        title = "Random Values"
        size = 800 to 400
    }
}
```

```kotlin
// Alternative bar chart
df.plot {
    bars {
        x(index)
        y(value)
    }
    layout {
        title = "Random Values - Bar Chart"
        size = 800 to 400
    }
}
```
